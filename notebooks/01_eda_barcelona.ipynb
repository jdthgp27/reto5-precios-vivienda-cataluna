```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import sys
import warnings

warnings.filterwarnings("ignore")

# Configuración de estilo
sns.set_theme(style="whitegrid", palette="viridis")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["figure.dpi"] = 100

# Rutas
BASE_DIR = Path(r"C:\Users\jdthg\Documents\curso6BI\reto5_precios_vivienda_cataluna")
DATA_PROCESSED = BASE_DIR / "data" / "processed"
FIGURAS_DIR = BASE_DIR / "reports" / "figuras"
FIGURAS_DIR.mkdir(parents=True, exist_ok=True)

print("✅ Configuración lista")
print(f"Figuras se guardarán en: {FIGURAS_DIR}")
```

```python
# Vista 1: total por barrio-mes
df = pd.read_csv(DATA_PROCESSED / "barcelona_barrio_mes.csv", parse_dates=["fecha"])

# Vista 2: por tipología
df_tipo = pd.read_csv(
    DATA_PROCESSED / "barcelona_barrio_mes_tipologia.csv", parse_dates=["fecha"]
)

print(f"Vista 1 (barrio × mes): {df.shape}")
print(f"Vista 2 (barrio × mes × tipología): {df_tipo.shape}")
print()
print("Rango temporal:", df["fecha"].min().date(), "→", df["fecha"].max().date())
print(f"Barrios únicos: {df['Nom_Barri'].nunique()}")
print(f"Distritos únicos: {df['Nom_Districte'].nunique()}")
```

```python
print("=== ESTADÍSTICAS DE total_transacciones ===")
print(df["total_transacciones"].describe().round(2))
print()
print("Percentiles:")
for p in [10, 25, 50, 75, 90, 95, 99]:
    v = df["total_transacciones"].quantile(p / 100)
    print(f"  P{p}: {v:.1f}")
print()
print(f"Mes con más transacciones en un barrio: {df['total_transacciones'].max()}")
print(f"Mes con menos transacciones en un barrio: {df['total_transacciones'].min()}")
```
